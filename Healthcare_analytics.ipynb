```python
df["Satisfaction_Score"] = df["Satisfaction_Score"].fillna(df["Satisfaction_Score"].median())
```

```python
df["Billing_Amount"] = df["Billing_Amount"].fillna(df["Billing_Amount"].median())
```

```python
df["Gender"].unique()
```

```python
df["Gender"] = df["Gender"].str.strip().str.upper()
```

```python
df["Gender"] = df["Gender"].replace({
    "M": "MALE",
    "F": "FEMALE",
    "MALE": "MALE",
    "FEMALE": "FEMALE"
})
```

```python
df["Department"].unique()
```

```python
df["Department"] = df["Department"].replace({
    "Cardio": "Cardiology",
    "General Med": "General Medicine"
})
```

```python
df["Diagnosis"].value_counts()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(data=df, x="Diagnosis", order=df["Diagnosis"].value_counts().index)

plt.xticks(rotation=45)
plt.title("Most Common Diagnoses")
plt.show()
```

```python
df.groupby("Department")["Billing_Amount"].mean().sort_values(ascending=False)
```

```python
df["Gender"].value_counts()
```

```python
df["Gender"].value_counts().plot.pie(
    autopct='%1.1f%%',
    figsize=(6,6)
)

plt.title("Patient Distribution by Gender")
plt.ylabel("")
plt.show()
```

```python
df["City"].value_counts().head(10)
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    y="City",
    order=df["City"].value_counts().index[:10]
)

plt.title("Top Cities by Patient Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df["Billing_Amount"], bins=30)

plt.title("Billing Amount Distribution")
plt.show()
```

```python
numeric_df = df.select_dtypes(include=["int64","float64"])
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
df.groupby("Diagnosis")["Length_of_Stay"].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=df.groupby("Diagnosis")["Length_of_Stay"].mean().sort_values().values,
    y=df.groupby("Diagnosis")["Length_of_Stay"].mean().sort_values().index
)

plt.title("Average Length of Stay by Diagnosis")

plt.show()
```

```python
#Insurance Provider Analysis
df["Insurance_Provider"].value_counts()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x="Insurance_Provider",
    order=df["Insurance_Provider"].value_counts().index
)

plt.title("Patient Distribution by Insurance Provider")

plt.xticks(rotation=45)

plt.show()
```

```python
#which insurance providers dominate patient admissions
df.groupby("Department")["Satisfaction_Score"].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=df.groupby("Department")["Satisfaction_Score"].mean().sort_values().values,
    y=df.groupby("Department")["Satisfaction_Score"].mean().sort_values().index
)

plt.title("Average Satisfaction Score by Department")

plt.show()
```

```python
#operational quality by department
plt.figure(figsize=(10,6))

sns.boxplot(x=df["Billing_Amount"])

plt.title("Billing Amount Outlier Detection")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df["Age"], bins=20)

plt.title("Patient Age Distribution")

plt.show()
```

```python
revenue_by_diagnosis = df.groupby("Diagnosis")["Billing_Amount"].sum().sort_values(ascending=False)

revenue_by_diagnosis
```

```python
plt.figure(figsize=(10,6))

revenue_by_diagnosis.plot(kind="bar")

plt.title("Revenue by Diagnosis")

plt.show()
```

```python
#Which Department Has Highest Revenue Per Patient?
dept_revenue = df.groupby("Department")["Billing_Amount"].mean().sort_values(ascending=False)

dept_revenue
```

```python
#Satisfaction vs Billing Analysis
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Billing_Amount",
    y="Satisfaction_Score"
)

plt.title("Billing Amount vs Satisfaction Score")

plt.show()
```

```python
#Top 10 Cities by Revenue
city_revenue = df.groupby("City")["Billing_Amount"].sum().sort_values(ascending=False).head(10)

city_revenue
```

```python
city_revenue.plot(kind="barh")

plt.title("Top Cities by Revenue")

plt.show()
```
